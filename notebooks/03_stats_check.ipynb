```python
import pandas as pd
from scipy.stats import chi2_contingency

df = pd.read_csv("netflix_customer_churn.csv")
df["inactive_30"] = df["last_login_days"] > 30

table = pd.crosstab(df["inactive_30"], df["churned"])
chi2, p_value, dof, expected = chi2_contingency(table)
print(table)
print("p-value:", p_value)
```

```python
df["age_group"] = pd.cut(df["age"], bins=[17, 24, 34, 44, 54, 64, 120],
                         labels=["18-24", "25-34", "35-44", "45-54", "55-64", "65+"])

for col in ["subscription_type", "payment_method", "region", "age_group"]:
    table = pd.crosstab(df[col], df["churned"])
    chi2, p, dof, expected = chi2_contingency(table)
    print(col, "| p-value:", round(p, 4))
```

```python
sub = df[df["subscription_type"].isin(["Standard", "Premium"])]
table = pd.crosstab(sub["subscription_type"], sub["churned"])
print("Standard vs Premium p-value:", round(chi2_contingency(table)[1], 4))
```
